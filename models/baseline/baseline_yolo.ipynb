```python
from ultralytics import YOLO
import ultralytics
```

```python
# Show YOLO version and confirm installation
print("Ultralytics version:", ultralytics.__version__)
ultralytics.checks()  # this is now a *module function*, not a class method
```

```python
# Load a small pretrained YOLO model (nano = light & fast)
model = YOLO("yolov8n.pt")

# Train on your snow pole dataset
results = model.train(
    data="./../../data.yaml",    # dataset definition file
    epochs=10,                 # small run to test everything works
    imgsz=640,                 # standard YOLO size
    batch=8,                   # adjust if you hit memory limits
    name="baseline_yolov8n",   # run name (creates runs/detect/baseline_yolov8n/)
    verbose=True
)
```

```python
results = model.predict(
    source="../../data/images/Test/test",
    save=True,
    conf=0.25
)

# Visualize a few example predictions
from IPython.display import Image, display
import glob

for image_path in glob.glob("runs/detect/predict/*.jpg")[:5]:
    display(Image(filename=image_path))
```
